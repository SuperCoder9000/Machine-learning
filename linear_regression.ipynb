## Linear Regression

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
```

```python
# Example: Study Hours vs Marks

x = np.array([1, 2, 3, 4, 5, 6]).reshape(-1, 1)
y = np.array([35, 45, 50, 55, 65, 70])
```

```python
model = LinearRegression()
model.fit(x, y)
```

```python
# Predict score for 7 hours of study
predicted_score = model.predict([[7]])
print("Predicted Score for 7 hours of study:", predicted_score[0])
```

```python
plt.scatter(x, y, color='blue', label="Actual Data")
plt.plot(x, model.predict(x), color='red', label="Regression Line")
plt.xlabel("Hours Studied")
plt.ylabel("Marks Scored")
plt.title("Linear Regression: Hours vs Marks")
plt.legend()
plt.show()
```
