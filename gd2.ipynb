## Gradient descent (using root mean square)

```python


import numpy as np


x = np.array([1, 2, 3, 4, 5], dtype=float)
y = np.array([5, 6, 10, 13, 11], dtype=float)
n = len(x)


B0_init = 3
B1_init = 1.9
lr = 0.1
epochs = 10

def predict(B0, B1, x):
    return B0 + B1 * x

def rmse(B0, B1, x, y):
    preds = predict(B0, B1, x)
    return np.sqrt(np.mean((y - preds)**2))

print("Initial B0:", B0_init)
print("Initial B1:", B1_init)
print("Learning rate:", lr)
```

```python

B0 = B0_init
B1 = B1_init
history_batch = []

for epoch in range(1, epochs + 1):
    preds = predict(B0, B1, x)
    errors = y - preds
    
    current_rmse = np.sqrt(np.mean(errors**2))
    
    # RMSE gradients
    grad_B0 = -(1 / (n * current_rmse)) * np.sum(errors)
    grad_B1 = -(1 / (n * current_rmse)) * np.sum(errors * x)
    
    B0 = B0 - lr * grad_B0
    B1 = B1 - lr * grad_B1
    
    loss = rmse(B0, B1, x, y)
    history_batch.append((epoch, B0, B1, loss))
    
    print(f"Epoch {epoch:2d} | B0 = {B0:.6f} | B1 = {B1:.6f} | RMSE = {loss:.6f}")

print("\nFinal Batch Gradient Descent Parameters:")
print("B0 =", B0, " | B1 =", B1)
```

```python


B0 = B0_init
B1 = B1_init
history_sgd = []

for epoch in range(1, epochs + 1):
    
    for i in range(n):
        xi = x[i]
        yi = y[i]
        
        pred_i = B0 + B1 * xi
        err_i = yi - pred_i
        
        current_rmse = rmse(B0, B1, x, y)
        
        grad_B0_i = -(1 / (n * current_rmse)) * err_i
        grad_B1_i = -(1 / (n * current_rmse)) * (err_i * xi)
        
        B0 = B0 - lr * grad_B0_i
        B1 = B1 - lr * grad_B1_i
    
    loss = rmse(B0, B1, x, y)
    history_sgd.append((epoch, B0, B1, loss))
    
    print(f"Epoch {epoch:2d} | B0 = {B0:.6f} | B1 = {B1:.6f} | RMSE = {loss:.6f}")

print("\nFinal Stochastic Gradient Descent Parameters:")
print("B0 =", B0, " | B1 =", B1)
```

```python

X = np.vstack([np.ones(n), x]).T
theta = np.linalg.inv(X.T.dot(X)).dot(X.T).dot(y)

B0_ols, B1_ols = theta[0], theta[1]
loss_ols = rmse(B0_ols, B1_ols, x, y)

print(f"OLS solution (MSE optimized) -> B0:{B0_ols:.6f}, B1:{B1_ols:.6f}, RMSE:{loss_ols:.6f}")
```
