## Multiple Linear Regression

```python
from sklearn.linear_model import LinearRegression
import numpy as np


print("Enter number of training samples: ")
n = int(input())

X_list = []
y_list = []

print("\nEnter data in format: size bedrooms floors age price")

for i in range(n):
    print(f"Sample {i+1}: ", end="")
    vals = list(map(float, input().split()))
    
   
    X_list.append(vals[:-1])
    y_list.append(vals[-1])

X = np.array(X_list, dtype=float)
y = np.array(y_list, dtype=float)


model = LinearRegression()
model.fit(X, y)

print("\nModel trained successfully!")
print("b0 (intercept) =", model.intercept_)
print("b-values (coefficients):", model.coef_)


print("\nEnter a new house to predict price:")
size     = float(input("Size (sq ft): "))
bedrooms = float(input("Bedrooms: "))
floors   = float(input("Floors: "))
age      = float(input("Age: "))

new_data = np.array([[size, bedrooms, floors, age]], dtype=float)

pred = model.predict(new_data)

print("\nPredicted Price =", round(pred[0], 2))
```
