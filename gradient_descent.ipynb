## Gradient Descent

```python
import numpy as np

# Data
x = np.array([1, 2, 3, 4, 5], dtype=float)
y = np.array([5, 6, 10, 13, 11], dtype=float)
n = len(x)

B0_init = 3.0
B1_init = 1.9
lr = 0.1
epochs = 10

def predict(B0, B1, x_arr):
    return B0 + B1 * x_arr

def mse(B0, B1, x_arr, y_arr):
    preds = predict(B0, B1, x_arr)
    return np.mean((y_arr - preds)**2)

print("Data x:", x)
print("Data y:", y)
print(f"Initial B0={B0_init}, B1={B1_init}, lr={lr}, epochs={epochs}")
```

```python
B0 = B0_init
B1 = B1_init

history_batch = []   
for epoch in range(1, epochs + 1):
    preds = predict(B0, B1, x)
    errors = y - preds
    grad_B0 = (-2.0 / n) * np.sum(errors)
    grad_B1 = (-2.0 / n) * np.sum(errors * x)
    B0 = B0 - lr * grad_B0
    B1 = B1 - lr * grad_B1
    loss = mse(B0, B1, x, y)
    history_batch.append((epoch, B0, B1, loss))
    print(f"Epoch {epoch:2d} | B0 = {B0:.6f} | B1 = {B1:.6f} | MSE = {loss:.6f}")

print("\nFinal batch params:", "B0 =", B0, "B1 =", B1)
```

```python
B0 = B0_init
B1 = B1_init

history_sgd = []  

for epoch in range(1, epochs + 1):
    
    for i in range(n):
        xi = x[i]
        yi = y[i]
        pred_i = B0 + B1 * xi
        err_i = yi - pred_i
        grad_B0_i = (-2.0 / n) * err_i
        grad_B1_i = (-2.0 / n) * err_i * xi
        B0 = B0 - lr * grad_B0_i
        B1 = B1 - lr * grad_B1_i
    loss = mse(B0, B1, x, y)
    history_sgd.append((epoch, B0, B1, loss))
    print(f"Epoch {epoch:2d} | B0 = {B0:.6f} | B1 = {B1:.6f} | MSE = {loss:.6f}")

print("\nFinal stochastic params:", "B0 =", B0, "B1 =", B1)
```

```python

X = np.vstack([np.ones(n), x]).T   
theta = np.linalg.inv(X.T.dot(X)).dot(X.T).dot(y)
B0_ols, B1_ols = theta[0], theta[1]
loss_ols = mse(B0_ols, B1_ols, x, y)
print(f"OLS solution -> B0 = {B0_ols:.6f} | B1 = {B1_ols:.6f} | MSE = {loss_ols:.6f}")
```
