## Web Scraping Example

```python
!pip install requests beautifulsoup4 pandas --quiet

import requests
from bs4 import BeautifulSoup
import pandas as pd

url = "https://quotes.toscrape.com/page/1/"
html = requests.get(url, headers={"User-Agent": "simple-scraper/1.0"}).text
soup = BeautifulSoup(html, "html.parser")

data = []
for q in soup.select("div.quote"):
    text = q.select_one("span.text").get_text(strip=True)
    author = q.select_one("small.author").get_text(strip=True)
    tags = ",".join([t.get_text(strip=True) for t in q.select(".tags .tag")])
    data.append({"text": text, "author": author, "tags": tags})

df = pd.DataFrame(data)
df
```

## Image Collection

```python
!pip install opencv-python pandas --quiet

import cv2
import os
import pandas as pd


folder = r"C:\Users\KIIT0001\Downloads"

image_data = []
image_names = []

for file in os.listdir(folder):
    if file.lower().endswith((".png", ".jpg", ".jpeg")):
        path = os.path.join(folder, file)
        img = cv2.imread(path)
        if img is not None:
            image_data.append(img)
            image_names.append(file)

print("Images loaded:", len(image_data))

pd.DataFrame({
    "filename": image_names,
    "shape (H,W,C)": [img.shape for img in image_data]
})
```

## Extracting from Text files

```python
import os
import pandas as pd


folder = r"C:\Users\KIIT0001\Downloads"

text_data = []
file_names = []

for file in os.listdir(folder):
    if file.lower().endswith(".txt"):
        path = os.path.join(folder, file)
        try:
            with open(path, "r", encoding="utf-8", errors="ignore") as f:
                text = f.read()
                text_data.append(text)
                file_names.append(file)
        except Exception as e:
            print(f"Error reading {file}:", e)

print("Text files loaded:", len(text_data))

# Display data as a table
pd.DataFrame({
    "filename": file_names,
    "content": text_data
})
```
